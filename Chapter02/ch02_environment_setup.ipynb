```python
from pathlib import Path

DATA_DIR = Path(r"C:\llm_data_analysis\chapter01\chapter01_RawData")

print("폴더 존재 여부:", DATA_DIR.exists())

files = [
    p for p in DATA_DIR.rglob("*")
    if p.is_file() and ("customer" in p.name.lower() or "고객" in p.name)
]

for f in files:
    print(f)
```

```python
import pandas as pd

customers_path = files[0]

if customers_path.suffix.lower() == ".csv":
    customers = pd.read_csv(customers_path)
elif customers_path.suffix.lower() in [".xlsx", ".xls"]:
    customers = pd.read_excel(customers_path)
else:
    raise ValueError(f"지원하지 않는 파일 형식입니다: {customers_path.suffix}")

print("사용 파일:", customers_path)
print("customers.shape:", customers.shape)
print("주요 컬럼:", customers.columns.tolist())

display(customers.head())
```
